Q1. Write a Python program to demonstrate multiple inheritance.
1. Employee class has 3 data members EmployeeID, Gender (String), Salary and
PerformanceRating(Out of 5) of type int. It has a get() function to get these details from
the user.
2. JoiningDetail class has a data member DateOfJoining of type Date and a function
getDoJ to get the Date of joining of employees.
3. Information Class uses the marks from Employee class and the DateOfJoining date
from the JoiningDetail class to calculate the top 3 Employees based on their Ratings
and then Display, using readData, all the details on these employees in Ascending
order of their Date Of Joining.


```python
from datetime import date

class employee:
    def __init__(self):
        self.employeeID=None
        self.gender=None
        self.salary=None
        self.performance=None

    def get(self):
        self.employeeID= int(input("enter your employee ID :"))
        self.gender = input("enter your gender:")
        self.salary = int(input("enter your salary"))
        self.performance= int(input("performance of the employee from 0 to 5:"))

class JoiningDetails:
    def __init__(self):
        self.DateofJoining= None

    def getDOJ(self):
        day = int(input("enter day of joining :"))
        month = int(input("enter month of joining :"))
        year = int(input("enter year of joining :"))
        self.DateofJoining = date(year,month,day)

class information(employee,JoiningDetails):
    #it is multiple inheritance which says one child having two or more parents
    def __init__(self):
        employee.__init__(self)
        JoiningDetails.__init__(self)

    def readData(self):
        print("enter employee detaills : ")
        self.get()
        self.getDOJ()

    def display(self):
        print(f"Employee ID       : {self.employeeID}")
        print(f"Gender            : {self.gender}")
        print(f"Salary            : {self.salary}")
        print(f"Performance Rating: {self.performance}/5")
        print(f"Date of Joining   : {self.DateofJoining.strftime('%d-%m-%Y')}")


def main():
    n = int(input("enter number of employee :"))
    employees = []
    for i in range(n):
        emp = information()
        emp.readData()
        emp.display()
        employees.append(emp)


    top3 = sorted(employees, key= lambda  e:e.performance, reverse=True)[:3] # [:3] means it'll take upto 2 indexing only
    #key and e is used to tell top3- what to sort (performance)
    #reverse = true means its making things right like we need top 3 from highest rating and sorted will do it in ascending

    top3_bydob = sorted(top3 ,key=lambda e:e.DateofJoining)

    print("Top 3 employees ( by rating ) , sorted by joining date :")

    for i in top3_bydob:
        i.display()


if __name__ == "__main__":
    main()
```

Q3. Consider an ongoing test cricket series. Following are the names of the players and their
scores in the test1 and 2.


Test Match 1 :
Dhoni : 56 , Balaji : 94


Test Match 2 :
Balaji : 80 , Dravid : 105
Calculate the highest number of runs scored by an individual cricketer in both of the matches.
Create a python function Max_Score (M) that reads a dictionary M that recognizes the player
with the highest total score. This function will return ( Top player , Total Score ) . You can
consider the Top player as String who is the highest scorer and Top score as Integer .
Input : Max_Score({‘test1’:{‘Dhoni’:56, ‘Balaji : 85}, ‘test2’:{‘Dhoni’ 87, ‘Balaji’’:200}})
Output : (‘Balaji ‘ , 200)

```python
class cricket:
    def __init__(self,matchs):
        self.matchs=matchs

    def max_score(self):
        totals={}

        for match in self.matchs.values(): #will give values onlly- not keys
            for player, runs in match.items(): #will give whole key value pair,gives you key-value pairs together, as tuples:
                #[('Dhoni', 56), ('Balaji', 94)]
                

#1st loop: player = 'Dhoni', runs = 56
#2nd loop: player = 'Balaji', runs = 94
                
                totals[player]= totals.get(player,0) + runs
#Search for the key 'Dhoni'. Return whatever number is stored next to it." It gives you 56 (a plain integer) — 
#not the text 'Dhoni' - thats what the get do...
#56 + 56    # if totals already had Dhoni:56 , and runs is 56
            
        top_player = max(totals, key=totals.get)
    #totals.get => gives us the value pares. 
            
        top_score = totals[top_player]

            
        return (top_player, top_score)



M = {
    'match1' : {'dhoni' : 56, 'balaji' : 94},
    'match2' : {'balaji' :80 , 'dravid' :105} 
}

series = cricket(M)
print(series.max_score())
```

Q4. Create a simple Card game in which there are 8 cards which are randomly chosen from a
deck. The first card is shown face up. The game asks the player to predict whether the next card
in the selection will have a higher or lower value than the currently showing card.
For example, say the card that’s shown is a 3. The player chooses “higher,” and the next card is
shown. If that card has a higher value, the player is correct. In this example, if the player had
chosen “lower,” they would have been incorrect. If the player guesses correctly, they get 20
points. If they choose incorrectly, they lose 15 points. If the next card to be turned over has the
same value as the previous card, the player is incorrect.

```python
import random
class game:
    def __init__(self):
        self.ranks=list(range(1,14))
        self.suits = ['♠', '♥', '♦', '♣']
        self.score=0
        self.hand=[]

    def create_hand(self):
        deck=[]
        for rank in self.ranks:
            for suit in self.suits:
                deck.append((rank,suit))
        random.shuffle(deck)
        self.hand= deck[:8] #we are taking 8 cards only

    def play(self):
        self.create_hand()
        current = self.hand[0] #"get the item at index 0" — i.e., the first card in the list of 8 eg- (7, '♥')
        print('card 1 :', current[0],current[1]) # current[0]=7 ,current[1]=♥

        for i in range(1,8):
            guess = input('Higher than this card or lower - (h/l) :').lower()
        # user types "H" -> guess becomes "h"
        # user types "l" -> guess stays "l
            next_card = self.hand[i]
            print(f"Card {i+1}:", next_card[0], next_card[1])
        
            if next_card[0] == current[0]:#cut marks when we get same numbers like 6 of hearts and 6 of spade
                    print("Same value, you lose points!")
                    self.score -= 15
            elif guess == "h" and next_card[0] > current[0]:
                    print("Correct!")
                    self.score += 20
            elif guess == "l" and next_card[0] < current[0]:
                    print("Correct!")
                    self.score += 20
            else:
                    print("Wrong!")
                    self.score -= 15

            current = next_card
    
        print("Final Score:", self.score)


gam2e = game()
gam2e.play()
        
```
