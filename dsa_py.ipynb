## Space Complexity
- is the amount of memory an algorithm uses as input size grows (input size matters only when input is using in the code - we defined 10 inputs but using also 5 in algo then the space complexity will be according to the 5 inputs only which was given in algorithm)

1. Big O - worst case
2. Big Omega - best case
3. Theta - average case

## Time Complexity
- is the amount of time an algorithm uses as input size grows
- It tells us how fast or slow is our algorithm is when data increases

Types of time complexity
1. O(1) -  constant time - always take the same time
2. O(n) - linear time - time grows with size
3. O(log n) - time shrinks because input is divided
4. O(n^2) - quadratic - time grows fast with nested loops
5. O(nlogn)- n logrithmic function- dividing sorting and merging etc

````python
#Lets cook this without the inbuild functions - insert(index, value)

#insertion 
def insert(arr,size,capacity,pos,value):
    # arr = array (as a list acting like a fixed-size array)
    # size = current number of elements in array
    # capacity = max size of array
    # pos = index where you want to insert (0-based)
    # value = element to insert
    if size>=capacity:
        print("You cant insert the array, its full")
        return size

    if pos<0 or pos>size:
        print("Invalid index")
        return size

        # Shift elements to the right, starting from the end
    i = size
    while i > pos: #O(n)
        arr[i] = arr[i - 1]
        i -= 1

    # Place the new value at pos
    arr[pos] = value

    # Increase size
    size += 1
    return size


capacity = 10
arr = [0] * capacity   # pre-allocated array of fixed capacity
arr[0], arr[1], arr[2] = 10, 20, 30
size = 3   # only 3 elements are "actually" in the array right now

print("Before insertion:", arr[:size])

size = insert(arr, size, capacity, 1, 99)   # insert 99 at index 1

print("After insertion:", arr[:size])
````

```
Singly Linked List — Definition & Key Points
==============================================

DEFINITION
----------
A Singly Linked List is a linear data structure where each element 
(called a NODE) contains:
  1. DATA  -> the actual value stored
  2. NEXT  -> a reference/pointer to the next node in the sequence

Unlike arrays, nodes are NOT stored in contiguous memory. Each node 
just "points" to the next one via its address. The last node's NEXT 
points to NULL, marking the end of the list.

STRUCTURE
---------
Head
 |
 v
+------+------+     +------+------+     +------+------+
| DATA | NEXT | --> | DATA | NEXT | --> | DATA | NEXT | --> NULL
+------+------+     +------+------+     +------+------+


KEY POINTS
----------
1. Traversal is one-directional only (head -> tail). You cannot go 
   backward like in a doubly linked list.

2. Head is a reference to the first node. If Head = None, the list 
   is empty.

3. Dynamic size — no fixed capacity like arrays. Grows/shrinks at 
   runtime by allocating/deallocating nodes.

4. Insertion/Deletion at the head = O(1), no shifting needed 
   (unlike arrays, where inserting at front = O(n)).

5. Insertion/Deletion at an arbitrary position = O(n), because you 
   must traverse from Head to reach that position first.

6. Random access (like arr[i]) is NOT possible in O(1). You must 
   traverse node-by-node -> O(n) access time.

7. Extra memory overhead per node (for storing the NEXT pointer), 
   compared to arrays which store only data.

8. Search = O(n) — no way to jump to a middle node directly.

9. No memory wastage from over-allocation (unlike arrays which may 
   reserve unused capacity).

10. Last node's NEXT is always None/NULL — this is how you detect 
    the end of the list while traversing.


TIME COMPLEXITY SUMMARY
------------------------
Operation                  Time
-----------------------    -----
Access (by index)          O(n)
Search                     O(n)
Insert at head              O(1)
Insert at tail (no tail ptr) O(n)
Insert at tail (with tail ptr) O(1)
Delete at head              O(1)
Delete at tail (no tail ptr) O(n)
```

```
Doubly Linked List — Definition & Key Points
===============================================

DEFINITION
----------
A Doubly Linked List is a linear data structure where each NODE 
contains THREE parts:
  1. PREV  -> reference/pointer to the previous node
  2. DATA  -> the actual value stored
  3. NEXT  -> reference/pointer to the next node

Unlike a singly linked list, you can traverse in BOTH directions 
(forward and backward). The first node's PREV points to NULL, and 
the last node's NEXT points to NULL.

STRUCTURE
---------
Head                                                          Tail
 |                                                              |
 v                                                              v
NULL <--+------+------+------+     +------+------+------+--> NULL
        | PREV | DATA | NEXT | <-> | PREV | DATA | NEXT |
        +------+------+------+     +------+------+------+
                  ^                          |
                  +------ <-- ----- <--------+


KEY POINTS
----------
1. Bidirectional traversal — can move forward (via NEXT) or 
   backward (via PREV), unlike singly linked list (forward only).

2. Each node stores TWO pointers instead of one -> more memory 
   overhead per node compared to singly linked list.

3. Deletion is easier/faster — you don't need to track the 
   "previous" node manually while traversing, since PREV is 
   already stored in the node itself.

4. First node's PREV = NULL, Last node's NEXT = NULL (these mark 
   the boundaries of the list).

5. Often maintains both a Head pointer AND a Tail pointer, making 
   insertion/deletion at the end O(1) as well (not just at head).

6. Insertion/Deletion at head or tail = O(1) (with tail pointer 
   maintained).

7. Insertion/Deletion at an arbitrary position = O(n) to reach 
   that position, but the actual insert/delete operation itself 
   is O(1) once you're there (just relinking PREV/NEXT).

8. Random access (like arr[i]) still NOT possible in O(1) — must 
   traverse node-by-node -> O(n).

9. Used in real-world structures like: browser back/forward 
   history, undo/redo functionality, LRU Cache, music playlists 
   (next/previous track).

10. More pointer operations = more chances of bugs during 
    insert/delete (must carefully update BOTH prev.next and 
    next.prev).


TIME COMPLEXITY SUMMARY
------------------------
Operation                     Time
---------------------------   -----
Access (by index)             O(n)
Search                        O(n)
Insert at head                 O(1)
Insert at tail (with tail ptr) O(1)
Delete at head                 O(1)
Delete at tail (with tail ptr) O(1)
Delete a given node (if you 
  already have reference to it) O(1)   <- key advantage over singly


SINGLY vs DOUBLY — QUICK COMPARE
----------------------------------
Feature              Singly LL        Doubly LL
------------------   --------------    --------------
Pointers per node     1 (next)          2 (prev, next)
Traversal direction   Forward only      Both directions
Memory per node        Less              More
Delete given node      O(n)              O(1)
Reverse traversal      Not possible      Possible
```\
