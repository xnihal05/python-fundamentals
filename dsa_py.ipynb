## Space Complexity
- is the amount of memory an algorithm uses as input size grows (input size matters only when input is using in the code - we defined 10 inputs but using also 5 in algo then the space complexity will be according to the 5 inputs only which was given in algorithm)

1. Big O - worst case
2. Big Omega - best case
3. Theta - average case

## Time Complexity
- is the amount of time an algorithm uses as input size grows
- It tells us how fast or slow is our algorithm is when data increases

Types of time complexity
1. O(1) -  constant time - always take the same time
2. O(n) - linear time - time grows with size
3. O(log n) - time shrinks because input is divided
4. O(n^2) - quadratic - time grows fast with nested loops
5. O(nlogn)- n logrithmic function- dividing sorting and merging etc

## Arrays
- fixed size
- store in contiguous memory
- They are actually homogeneous - but in python if you'll take diff datatypes in array then all will be created into string!
- Fast access O(1) using index
- Insertion/Deletion is slow O(n) - number of elements

```python
a=[10,20,30,20,10,30]
for i in a: #O(n)
    if i==30:
        print("Access using traversal :", i, 'index :',a.index(i) )
        break

#Time complexity for this algo is O(n)

print("accessing element using index", a[2]) #O(1) - time complexity
```

```python
a=[10,20,30,20,10,30]
#insert - traverse swap and insert
a.append(100) #it will add value at the end - o(n)
a.insert(3,500) #it will insert at the  3rd indexing - o(n)
print(a)
#replace - o(n)
a[2]=200
print (a)

#deletion - o(n) - traverse delete and swap
a.pop() #will delete the last element - O(1)
print(a)
a.remove(20) #will remove the 20 which will come first - but its not for array 
print (a)
a.pop(2) #will pop an index which is given
print(a)
a.pop(4) #O(1) - because its a last element so there is no swapping
print(a)
```

```python
#Lets cook this without the inbuild functions - insert(index, value)

#insertion 
def insert(arr,size,capacity,pos,value):
    # arr = array (as a list acting like a fixed-size array)
    # size = current number of elements in array
    # capacity = max size of array
    # pos = index where you want to insert (0-based)
    # value = element to insert
    if size>=capacity:
        print("You cant insert the array, its full")
        return size

    if pos<0 or pos>size:
        print("Invalid index")
        return size

        # Shift elements to the right, starting from the end
    i = size
    while i > pos:
        arr[i] = arr[i - 1]
        i -= 1

    # Place the new value at pos
    arr[pos] = value

    # Increase size
    size += 1
    return size


capacity = 10
arr = [0] * capacity   # pre-allocated array of fixed capacity
arr[0], arr[1], arr[2] = 10, 20, 30
size = 3   # only 3 elements are "actually" in the array right now

print("Before insertion:", arr[:size])

size = insert(arr, size, capacity, 1, 99)   # insert 99 at index 1

print("After insertion:", arr[:size])
```

```python
def append(arr,size,capacity, value):
    if size>= capacity:
        print("Size is full, you cant add the element in array")
        return size

    arr[size]= value
    size+=1
    return size

    print("adding value at last which is", value)

capacity = 10
size = 4
arr= [0]*capacity
arr=[2,121,1,43,45,73,32]
print("Before append:", arr[:size])

size = append(arr, size, capacity, 50)

print("After append:", arr[:size])
```

```python
def del_element(arr,size,pos):
    if size<=0:
        print("Array is empty,cannot delete any element")
        return size

    if pos<0 or pos>size:
        print("Wrong pos added to be delete")
        return size

    removed_value= arr[pos]
    print("Deleted value is", removed_value)

    i=pos
    while i<size-1:
        arr[i]=arr[i+1]
        i+=1

    size-=1
    return size


capacity = 10
arr = [0] * capacity
arr[0], arr[1], arr[2], arr[3] = 10, 99, 20, 30
size = 4

print("Before deletion:", arr[:size])

size = del_element(arr, size, 1)   # delete element at index 1 (value 99)

print("After deletion:", arr[:size])
```
