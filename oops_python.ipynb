# Inheritance 
When you inherit some properties from parent class- child class can use properties and method of parent's class
Make code more clean and manageable

Types :
1. Single
2. Multiple
3. Multi Level
4. Hierarchical

### Single - One Parent and One child (parent - child)
### Multi Level - Two parents(grandparent and parent) and two child classes(child of grandparent and parent) - simply multiple  & more than one parent
### Multiple - More than 1 parent and a single child ( Mother- Father- Child)
### Hierarchical - Single parent have multiple child ( Parent - brother- sister)

```python
#1. Single Inheritance

class parent:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def display(self):
        print(self.name,self.age)

class child(parent):
    def show(self):
        print(self.name,self.age)

obj = child('Nihal',19)
obj.show()
objw = parent('Meow',90)
objw.display()
```

```python
#2. Multi Level inheritance (2 parents , 2 child)

class grandparent:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def show(self):
        print(self.name,self.age)
class parent(grandparent):
    def __init__(self,name,age,profession):
        grandparent.__init__(self,name,age)
        self.profession=profession
    def display(self):
        print(self.name,self.age,self.profession)
class child(parent):
    def __init__(self,name,age,profession,degree_name):
        parent.__init__(self,name,age,profession)
        self.degree_name=degree_name
    def intro(self):
        print(self.name,self.age,self.degree_name)

obj=grandparent('ABC',90)
obj.show()
obj2=parent('DEF',44,'Cook')
obj2.display()
obj3=child('Nihal',19,'Student','BTECH')
obj3.intro()

#whenever every classes having their own constructer then we have to define all parameters in each constructor
```

```python
#multiple inheritance
class student:
    def __init__(self,name,student_id):
        self.name=name
        self.student_id=student_id

    def display(self):
        print(self.name,self.student_id)

class teacher:
    def __init__(self,subject):
        self.subject=subject
        

    def show(self):
        print(self.subject)

class teacher_assistant(student,teacher):
    def __init__(self,name,student_id,subject):
      teacher.__init__(self,subject)
      student.__init__(self,name,student_id)

    def show(self):
        print(self.name,self.student_id,self.subject)

obj= teacher_assistant("manav",445,"CSE")
obj.show()
obj34=teacher('AIML')
obj34.show()
```

```python
class bankaccount:
    def __init__(self,account_holder):
        self.account_holder= account_holder
        self.__balance=0 #private
        self._transaction_logs=[] #protected

    def deposite(self,amount):
        self.amount=amount
        if amount>0:
            self.__balance += self.amount
            self._transaction_logs.append(self.amount)

        else:
            print('invalid diposited amount')

    def get_balance(self):
        print(self.__balance)

    def show_log(self):
        print(self._transaction_logs)

obj=bankaccount('arjun')
obj.deposite(5000)
obj.get_balance()
obj.deposite(10000)
obj.get_balance()
obj.show_log()
```

```python
#you are designing a car rental system
#each car has a public brand name
#odometer cant be directly accessed or modified by users
#service history can be accessible to engineers 

class car:
    def __init__(self,brand,model):
        self.brand=brand #public
        self.model=model #public
        self.__odometer=0 #private
        self._service=[] #protected

    def drive(self,kms):
        if kms>0:
            self.__odometer += kms
            print(f'{self.brand} driven for {kms} kms')
        else:
            print('invalid distance')

    def get_odometer(self):
        return self.__odometer

    def service(self,details):
        self.details=details
        self._service.append(details)

    def service_show(self):
        return self._service


class service_engineer(car):
    def update_service(self,service_details):
        self._service.append(service_details)
        print('service log update by an engineer')

car1=car('bmw', 9988)
car1.drive(300)
car1.get_odometer()
car1.service('changed tires')
car1.service_show()
car2= service_engineer('hi',5)
car2.update_service('idk')
car2.service_show()
```

# Abstraction
Hiding the internal details and shows the necessary one only
Example - you are using a computer without knowing about how it runs on binary 

```python
from abc import ABC, abstractmethod
class vehcile(ABC):
    @abstractmethod #we hiding the complexity
    def start_engine(self):
        print('the fuel is burning and engine is moving back and forth with blasts')

#child class will always provide relevent information about the hidden function
class bike(vehcile):
    def start_engine(self):
        print('bike engine started')

obj= bike()
obj.start_engine()
#obj2= vehcile() - it will give the error cause it is hidden complexity
#obj2.start_engine() - it will give the error cause it is hidden complexity
```

```python
from abc import ABC , abstractmethod
class FoodDelivery(ABC):
    @abstractmethod
    def place_order(self,food_items):
        print('nothing')
    @abstractmethod
    def track_order(self,order_id):
        print('order id number is being processed')

class zomato(FoodDelivery):
    def place_order(self,food_items):
        print('zomato order placed for ', food_items)

    def track_order(self,order_id):
        print('zomato track id is', order_id)

obj=zomato()
obj.place_order('pizza')
obj.track_order(990)
```

```python
class controler(ABC):
    @abstractmethod
    def on(self):
        print('current is travelling from +ve to -ve')
    @abstractmethod
    def off(self):
        print('no movements of electrones')

class light(controler):
    def on(self):
        print('lights on')
    def off(self):
        print('lights off')

class fan(controler):
    def on(self):
        print('fan on')
    def off(self):
        print('fan off')

class ac(controler):
    def on(self):
        print('ac on')
    def off(self):
        print('ac off')

f=fan()
f.off()
a=ac()
a.on()
l=light()
l.off()
```

```python
#use ABC module to create an abstract class device with abstract method start()
#create subclass laption and class that implements start()

from abc import ABC , abstractmethod
class device(ABC):
    @abstractmethod
    def start(self):
        print('Internal things working in device')

class laptop(device):
    def start(self):
        print('welcome, windows are loading')

class phone(device):
    def start(self):
        print('Android Running')

#obj=device() - cant access that cause its in abstract method
obj = laptop()
obj.start()
obj2=phone()
obj2.start()
```

### Magic Methods
In pythod, they are special method starts and ends with double underscore
these methods gets automatically called when use some builtin operations (print,add,get,length)

- Why we use magic methods
1. _ _ s t r _ _ --> for string representation
used when print an object using print() or str()



```python
#1. __str__ --> for string representation

class book:
    def __init__(self,title,author):
        self.title=title
        self.author=author

    def __str__(self):
        return f'{self.title} by {self.author}'

my_book = book ('abc','idk')
print(my_book)
```

```python
#__len__ --> it is used when you call length in an object

class playlist:
    def __init__(self,songs):
        self.songs=songs

    def __len__(self):
        return len(self.songs)

my_playlist = playlist(['saih','fas','das'])
len(my_playlist)
```
