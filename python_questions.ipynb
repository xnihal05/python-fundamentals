Q8 Count vowels, consonants, digits, and spaces in a string.


```python
def count(string):
    vowels=0
    digits=0
    cons=0
    spaces=0
    for i in string:
        if i in 'aeiou':
            vowels+=1
        elif i.isdigit():
            digits+=1
        elif i==" ":
            spaces+=1
        elif i.isalpha(): #check if alphabet - also include vowels
            cons+=1

    return vowels, digits, cons,spaces

vowels, digits, cons,spaces=count("hello my name is nihal singh and i am 20 years young")
print(vowels)  
print(digits)
print(cons)
print(spaces)
```

Q9 Given a sentence, capitalize the first letter of every word without using .title().

```python
def capitalize(sentence):
    result=''
    for i in range(len(sentence)):
        if i==0 or sentence[i-1]==" ":
            result += sentence[i].upper()
        else:
            result+= sentence[i]

    return result

answer = capitalize('hi i am 20 years old And i am very happy')
print(answer)
```

Q10 : Remove all duplicate characters from a string while preserving order.

```python
def rem_dup(string):
    result=""
    seen=[]
    for char in string:
        if char not in seen:
            result+=char
            seen.append(char)

    return result

ans = rem_dup("nihal nihal")
print(ans)
```

```python
#set approach
def rem_dup(string):
    result=""
    seen=set() #makes it a set - we cant use seen={} it'll make it dictionary
    for char in string:
        if char not in seen:
            result+=char
            seen.add(char) #add in set only

    return result

ans = rem_dup("nihal nihal")
print(ans)

#its time complexity is low cause sets are using so that no duplicate values
```

Q11: Check if a string is a palindrome (ignore case and spaces).

A string that reads the same forward and backward. E.g. "madam" — reverse it, you still get "madam". But "hello" reversed is "olleh" — different, so not a palindrome.

Case shouldn't matter. "Madam" should still count as a palindrome, even though M ≠ m technically. So before comparing, you'd want to normalize everything to the same case — usually lowercase everything.

Spaces shouldn't count either. Something like "nurses run" — if you read it as pure letters (nursesrun), is that a palindrome? Spaces get treated as noise, not real characters. So you'd strip spaces out before checking.

```python
def palin(string):
    clean = string.lower().replace(" ","")
    if clean==clean[::-1]:
        return True
    else:
        return False
palin('Madam')
```

Q13: Unique elements, preserve order

```python
def unique_preserve(lists):
    seen= set()
    result= []
    for i in lists:
        if i not in seen:
            result.append(i)
            seen.add(i)
    return result

print(unique_preserve([2,4,4,4,5,6,7,6,7,6,7,8,8,9,7,4,4,3]))
```

Q14: Top 3 word frequency in a paragraph

```python
from collections import Counter

def top_3(para):
    words= para.lower().split() #.split() — breaks the string into a list of words on whitespace, given in below (example)
    words= [w.strip('",.!/\'') for w in words] 
        #"  hello  ".strip()        # "hello"  — removes leading/trailing whitespace by default
#"dog.".strip('.')          # "dog"    — removes '.' from the ends
#"...hi...".strip('.')      # "hi"     — removes ALL leading/trailing dots, however many
    counts= Counter(words) #will count the same words which we have filtered
    return counts.most_common(3)

para = "the quick brown fox jumps over the lazy dog. The dog barks, and the fox runs."
print(top_3(para))
```

Q15: is_prime(n), print primes below 100

```python
def is_prime(n):
    if n<2:
        return False
    for i in range(2,int (n**0.5)+1): #try dividing n by every number from 2 up to √n.
        #2**3    # 8   → 2 to the power of 3
#5**2    # 25  → 5 squared
#9**0.5  # 3.0 → 9 to the power of 0.5
        #Why int(...) around it?

#n**0.5 gives you a float (decimal), even when the answer is mathematically a whole number
#int(...) truncates the decimal part — chops it off, no rounding:
#Why + 1 at the end?
#This is about how range() works — range(2, x) stops before reaching x, it doesn't include it:
        if n%i ==0:
            return False
    return True

for num in range(100):
    if is_prime(num):
        print(num,end=' ')
```

Q16: Factorial


```python
def fac(n):
    if n < 0:
        raise ValueError("factorial not defined for negative numbers")
    if n == 0 or n == 1:
        return 1
    return n * fac(n - 1)

    
fac(6)
```
